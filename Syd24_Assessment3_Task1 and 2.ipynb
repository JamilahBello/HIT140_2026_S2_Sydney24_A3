## Accessing the FIFA Match Data

The FIFA match data is accessed through a browser session using Playwright.

The FIFA International Match Archive page is first opened in Chrome. A request to FIFA's match data service is then made from inside that browser session.

Reviewed the Playwright documentation to understand how to:

- launch Chromium
- create a browser context
- open a page
- use `page.evaluate()` to run JavaScript inside the page

The FIFA request itself uses the JavaScript `fetch()` function, so I also reviewed the MDN Fetch API tutorial.

The response is returned in JSON format and is later converted into a pandas DataFrame.

**Tutorials:**

- [Playwright BrowserType.launch](https://playwright.dev/python/docs/api/class-browsertype#browser-type-launch)
- [Playwright Page](https://playwright.dev/python/docs/api/class-page)
- [MDN: Using the Fetch API](https://developer.mozilla.org/en-US/docs/Web/API/Fetch_API/Using_Fetch)
- [Python json](https://docs.python.org/3/library/json.html)

```python
# Browser launch and page references:
# Microsoft (n.d.-a, n.d.-b)

# Fetch API reference:
# Mozilla (n.d.)
async def get_world_cup_matches():
  async with async_playwright() as p:

    # Launch the Google chrome installation available in the notebook environment
    browser = await p.chromium.launch(
        executable_path="/usr/bin/google-chrome",
        headless=True,
        args=["--no-sandbox", "--disable-dev-shm-usage"],
    )

    context = await browser.new_context(locale="en-AU")
    page = await context.new_page()

    # Open FIFA first to establish FIFA browser session
    await page.goto(
        "https://inside.fifa.com/data-centre/matches",
        wait_until="domcontentloaded",
        timeout=120000
    )

    await page.wait_for_timeout(5000)

    # Request 2026 FIFA World Cup matches
    url = ("https://inside.fifa.com/api/data-centre/matches?gender=1&language=en&count=200&year=2026&competitionClassificationCode=FWC")

    response = await page.evaluate(
      """
      async (url) => {
        const response = await fetch(url);

        return {
          status: response.status,
          body: await response.json(),
        };
      }
      """,
      url
    )

    await browser.close()

  # Check whether FIFA returned the request successfully.
  print("HTTP status:", response["status"])

  if response["status"] != 200:
    raise RuntimeError(f"Failed with HTTP status {response['status']}")

  return response["body"]

# Run the asynchronous function inside the notebook.
loop = asyncio.get_event_loop()
world_cup_raw = loop.run_until_complete(get_world_cup_matches())

print("Records returned:", len(world_cup_raw))
```

```python
# pandas DataFrame reference:
# pandas development team (n.d.-a)

def get_english_name(name_data):
  if not isinstance(name_data, list):
    return None

  for item in name_data:
    if item.get("locale") in ["en-GB", "en"]:
      return item.get("description")

  return None

world_cup_matches = []

for match in world_cup_raw:
  world_cup_matches.append({
      "MatchID": match.get("idMatchIfes"),
      "MatchDate": match.get("matchDate"),
      "TeamA": get_english_name(match.get("teamAName")),
      "TeamB": get_english_name(match.get("teamBName")),
      "TeamAID": match.get("teamAId"),
      "TeamBID": match.get("teamBId"),
      "TeamAGoals": match.get("teamAScore"),
      "TeamBGoals": match.get("teamBScore"),
      "Stage": get_english_name(match.get("stageName")),
      "HasPenalties": match.get("hasPenalties")
  })

matches_df = pd.DataFrame(world_cup_matches)

# Convert MatchDate from text into pandas datetime format.
matches_df["MatchDate"] = pd.to_datetime(
    matches_df["MatchDate"]
)

matches_df.head()
```

## Definition of Prior-Match Data

Prior-match data means information that was available before the kickoff of the particular World Cup match being predicted.

The definition is based on time rather than calendar year.

For example, if a World Cup match was played on 25 June 2026:

- a match played in December 2025 is prior-match information
- a match played on 5 June 2026 is prior-match information
- an earlier World Cup match played on 18 June 2026 is also prior-match information
- a match played on 28 June 2026 is not prior-match information

This rolling definition prevents future information from being used to predict an earlier match.

## Collecting Prior International Match Data

FIFA's International Match Archive is queried using each team's FIFA ID for matches from 2025 and 2026.

The same international match may be returned when searching for both participating teams, duplicate match records will later be removed.

**Tutorials:**

- [Playwright Page.evaluate](https://playwright.dev/python/docs/api/class-page#page-evaluate)
- [MDN: Using the Fetch API](https://developer.mozilla.org/en-US/docs/Web/API/Fetch_API/Using_Fetch)
- [Python asyncio](https://docs.python.org/3/library/asyncio.html)

```python
# Browser and fetch references:
# Microsoft (n.d.-b); Mozilla (n.d.)

async def get_prior_matches():

    all_matches = []

    async with async_playwright() as p:

        browser = await p.chromium.launch(
            executable_path="/usr/bin/google-chrome",
            headless=True,
            args=["--no-sandbox", "--disable-dev-shm-usage"]
        )

        context = await browser.new_context(locale="en-AU")
        page = await context.new_page()

        # Open FIFA first to establish a browser session.
        await page.goto(
            "https://inside.fifa.com/data-centre/matches",
            wait_until="domcontentloaded",
            timeout=120000
        )

        await page.wait_for_timeout(5000)

        # Retrieve recent matches for each World Cup team.
        for _, team_row in world_cup_teams.iterrows():

            team_id = str(team_row["TeamID"])

            team_name = (team_row["Team"])

            print(f"Retrieving: {team_name}")

            for year in [2025, 2026]:

                url = (f"https://inside.fifa.com/api/data-centre/matches?gender=1&language=en&count=100&teamId={team_id}&year={year}")

                response = await page.evaluate(
                    """
                    async (url) => {

                        const response =
                            await fetch(url);

                        return {
                            status:
                                response.status,
                            body:
                                await response.json()
                        };
                    }
                    """,
                    url
                )

                if response["status"] != 200:
                    raise RuntimeWarning(f"Warning with HTTP status {response['status']} for {team_name} ({year})")
                    continue

                team_matches = (response["body"])

                # Add information showing which team caused the record to be retrieved.
                for match in team_matches:
                    match[ "_RequestedTeam"] = team_name
                    match["_RequestedTeamID"] = team_id

                all_matches.extend( team_matches)

        await browser.close()

    return all_matches


loop = asyncio.get_event_loop()
prior_matches_raw = (loop.run_until_complete(get_prior_matches()))

print("\nRaw records retrieved:", len(prior_matches_raw))
```

## Checking Available Prior-Match History

The planned recent-form measures will use each team's five most recent matches before a World Cup match.

Before using a five-match window, the available history is checked to make sure every World Cup team has at least five earlier matches available.

This check is important because the same method should be possible for every team.

```python
world_cup_team_ids = pd.concat([
    matches_df[["TeamAID", "TeamA"]].rename(columns={"TeamAID": "TeamID", "TeamA": "Team"}),
    matches_df[["TeamBID", "TeamB"]].rename(columns={"TeamBID": "TeamID", "TeamB": "Team"})
]).drop_duplicates()

first_world_cup_match = pd.concat([
    matches_df[["TeamAID", "MatchDate"]].rename(columns={"TeamAID": "TeamID"}),
    matches_df[["TeamBID", "MatchDate"]].rename(columns={"TeamBID": "TeamID"})
]).groupby("TeamID")["MatchDate"].min()

history_counts = []

for _, team in world_cup_team_ids.iterrows():
    team_id = team["TeamID"]
    first_match_date = first_world_cup_match.loc[team_id]

    available_matches = team_history_df[
        (team_history_df["TeamID"].astype(str) == str(team_id)) &
        (team_history_df["MatchDate"] < first_match_date)
    ]

    history_counts.append({
        "Team": team["Team"],
        "AvailablePriorMatches": len(available_matches)
    })

history_counts_df = pd.DataFrame(history_counts).sort_values(
    "AvailablePriorMatches"
)

display(history_counts_df)
```

## Creating a Pool of Candidate Pre-Match Variables

Each candidate variable is calculated from a team's five most recent matches before the target World Cup match.

Several reasonable pre-match measures are created so they can later be compared using exploratory analysis.

The candidate variables include measures of:

- recent results
- attacking performance
- defensive performance
- scoring consistency
- defensive consistency
- recovery time

```python
def get_previous_five_features(team_id, target_date, history_df):
    previous_matches = (
        history_df[
            (history_df["TeamID"].astype(str) == str(team_id)) &
            (history_df["MatchDate"] < target_date)
        ]
        .sort_values("MatchDate")
    )

    last_five = previous_matches.tail(5)

    if len(last_five) < 5:
        return None

    most_recent_match_date = last_five["MatchDate"].max()

    return {
        "WinRate": last_five["Win"].mean(),
        "LossRate": last_five["Loss"].mean(),
        "DrawRate": last_five["Draw"].mean(),
        "AvgGoalsScored": last_five["GoalsFor"].mean(),
        "AvgGoalsConceded": last_five["GoalsAgainst"].mean(),
        "AvgGoalDifference": last_five["GoalDifference"].mean(),
        "CleanSheetRate": (last_five["GoalsAgainst"] == 0).mean(),
        "ScoringRate": (last_five["GoalsFor"] > 0).mean(),
        "TwoPlusGoalsRate": (last_five["GoalsFor"] >= 2).mean(),
        "ConcedeTwoPlusRate": (last_five["GoalsAgainst"] >= 2).mean(),
        "GoalsScoredStd": last_five["GoalsFor"].std(),
        "GoalsConcededStd": last_five["GoalsAgainst"].std(),
        "DaysRest": (target_date - most_recent_match_date).days
    }
```

### Candidate Variables Created

The function creates the following possible pre-match measures:

- `WinRate` – proportion of the previous five matches won
- `LossRate` – proportion of the previous five matches lost
- `DrawRate` – proportion of the previous five matches drawn
- `AvgGoalsScored` – average goals scored
- `AvgGoalsConceded` – average goals conceded
- `AvgGoalDifference` – average goal difference
- `CleanSheetRate` – proportion of matches where no goal was conceded
- `ScoringRate` – proportion of matches where the team scored at least once
- `TwoPlusGoalsRate` – proportion of matches where the team scored two or more goals
- `ConcedeTwoPlusRate` – proportion of matches where the team conceded two or more goals
- `GoalsScoredStd` – variation in goals scored
- `GoalsConcededStd` – variation in goals conceded
- `DaysRest` – number of days since the team's most recent match

```python
candidate_rows = []

for _, match in matches_df.iterrows():
    team_a_features = get_previous_five_features(
        match["TeamAID"],
        match["MatchDate"],
        team_history_df
    )

    team_b_features = get_previous_five_features(
        match["TeamBID"],
        match["MatchDate"],
        team_history_df
    )

    candidate_rows.append({
        "MatchID": match["MatchID"],
        "MatchDate": match["MatchDate"],
        "TeamA": match["TeamA"],
        "TeamB": match["TeamB"],

        "TeamAWinRate": team_a_features["WinRate"],
        "TeamALossRate": team_a_features["LossRate"],
        "TeamADrawRate": team_a_features["DrawRate"],
        "TeamAAvgGoalsScored": team_a_features["AvgGoalsScored"],
        "TeamAAvgGoalsConceded": team_a_features["AvgGoalsConceded"],
        "TeamAAvgGoalDifference": team_a_features["AvgGoalDifference"],
        "TeamACleanSheetRate": team_a_features["CleanSheetRate"],
        "TeamAScoringRate": team_a_features["ScoringRate"],
        "TeamATwoPlusGoalsRate": team_a_features["TwoPlusGoalsRate"],
        "TeamAConcedeTwoPlusRate": team_a_features["ConcedeTwoPlusRate"],
        "TeamAGoalsScoredStd": team_a_features["GoalsScoredStd"],
        "TeamAGoalsConcededStd": team_a_features["GoalsConcededStd"],
        "TeamADaysRest": team_a_features["DaysRest"],

        "TeamBWinRate": team_b_features["WinRate"],
        "TeamBLossRate": team_b_features["LossRate"],
        "TeamBDrawRate": team_b_features["DrawRate"],
        "TeamBAvgGoalsScored": team_b_features["AvgGoalsScored"],
        "TeamBAvgGoalsConceded": team_b_features["AvgGoalsConceded"],
        "TeamBAvgGoalDifference": team_b_features["AvgGoalDifference"],
        "TeamBCleanSheetRate": team_b_features["CleanSheetRate"],
        "TeamBScoringRate": team_b_features["ScoringRate"],
        "TeamBTwoPlusGoalsRate": team_b_features["TwoPlusGoalsRate"],
        "TeamBConcedeTwoPlusRate": team_b_features["ConcedeTwoPlusRate"],
        "TeamBGoalsScoredStd": team_b_features["GoalsScoredStd"],
        "TeamBGoalsConcededStd": team_b_features["GoalsConcededStd"],
        "TeamBDaysRest": team_b_features["DaysRest"]
    })

candidate_form_df = pd.DataFrame(candidate_rows)

display(candidate_form_df.head(10))
```

## Creating Team Difference Variables

For each candidate feature, a difference is created using:

**Team A value - Team B value**

A positive value means Team A has the higher recent value.

A negative value means Team B has the higher recent value.

```python
candidate_form_df["WinRateDifference"] = (candidate_form_df["TeamAWinRate"] - candidate_form_df["TeamBWinRate"])
candidate_form_df["LossRateDifference"] = (candidate_form_df["TeamALossRate"] - candidate_form_df["TeamBLossRate"])
candidate_form_df["DrawRateDifference"] = (candidate_form_df["TeamADrawRate"] - candidate_form_df["TeamBDrawRate"])
candidate_form_df["AvgGoalsScoredDifference"] = (candidate_form_df["TeamAAvgGoalsScored"] - candidate_form_df["TeamBAvgGoalsScored"])
candidate_form_df["AvgGoalsConcededDifference"] = (candidate_form_df["TeamAAvgGoalsConceded"] - candidate_form_df["TeamBAvgGoalsConceded"])
candidate_form_df["AvgGoalDifferenceDifference"] = (candidate_form_df["TeamAAvgGoalDifference"] - candidate_form_df["TeamBAvgGoalDifference"])
candidate_form_df["CleanSheetRateDifference"] = (candidate_form_df["TeamACleanSheetRate"] - candidate_form_df["TeamBCleanSheetRate"])
candidate_form_df["ScoringRateDifference"] = (candidate_form_df["TeamAScoringRate"] - candidate_form_df["TeamBScoringRate"])
candidate_form_df["TwoPlusGoalsRateDifference"] = (candidate_form_df["TeamATwoPlusGoalsRate"] - candidate_form_df["TeamBTwoPlusGoalsRate"])
candidate_form_df["ConcedeTwoPlusRateDifference"] = (candidate_form_df["TeamAConcedeTwoPlusRate"] - candidate_form_df["TeamBConcedeTwoPlusRate"])
candidate_form_df["GoalsScoredStdDifference"] = (candidate_form_df["TeamAGoalsScoredStd"] - candidate_form_df["TeamBGoalsScoredStd"])
candidate_form_df["GoalsConcededStdDifference"] = (candidate_form_df["TeamAGoalsConcededStd"] - candidate_form_df["TeamBGoalsConcededStd"])
candidate_form_df["DaysRestDifference"] = (candidate_form_df["TeamADaysRest"] - candidate_form_df["TeamBDaysRest"])
```

```python
candidate_columns = [
    "WinRateDifference",
    "LossRateDifference",
    "DrawRateDifference",
    "AvgGoalsScoredDifference",
    "AvgGoalsConcededDifference",
    "AvgGoalDifferenceDifference",
    "CleanSheetRateDifference",
    "ScoringRateDifference",
    "TwoPlusGoalsRateDifference",
    "ConcedeTwoPlusRateDifference",
    "GoalsScoredStdDifference",
    "GoalsConcededStdDifference",
    "DaysRestDifference"
]

candidate_analysis_df = matches_df[
    ["MatchID", "MatchDate", "TeamA", "TeamB", "GoalDifference"]
].merge(
    candidate_form_df[["MatchID"] + candidate_columns],
    on="MatchID",
    how="left"
)

print("Rows:", len(candidate_analysis_df))
print("Candidate variables:", len(candidate_columns))
print("Missing values:", candidate_analysis_df[candidate_columns].isna().sum().sum())

display(candidate_analysis_df.head(10))
```

```python
import asyncio
import nest_asyncio
from playwright.async_api import async_playwright
nest_asyncio.apply()

BASE = "https://inside.fifa.com"

def english(name_data):
    if not isinstance(name_data, list):
        return None
    for item in name_data:
        if item.get("locale") in ["en-GB", "en"]:
            return item.get("description")
    return None

async def fifa_fetch(page, url):
    response = await page.evaluate(
        "async (url) => { const r = await fetch(url); return {status: r.status, body: await r.json()}; }",
        url)
    if response["status"] != 200:
        raise RuntimeError(f"HTTP {response['status']} for {url}")
    return response["body"]

async def collect():
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True, args=["--no-sandbox"])
        page = await (await browser.new_context(locale="en-AU")).new_page()
        await page.goto(BASE + "/data-centre/matches", wait_until="domcontentloaded", timeout=120000)
        await page.wait_for_timeout(5000)

        wc = await fifa_fetch(page, BASE + "/api/data-centre/matches?gender=1&language=en&count=200"
                                          "&year=2026&competitionClassificationCode=FWC")
        team_ids = sorted({m["teamAId"] for m in wc} | {m["teamBId"] for m in wc})
        prior = []
        for tid in team_ids:
            for year in [2025, 2026]:
                prior += await fifa_fetch(page, BASE + f"/api/data-centre/matches?gender=1&language=en"
                                                        f"&count=100&teamId={tid}&year={year}")
        ranking = await fifa_fetch(page, BASE + "/api/ranking-overview?locale=en&dateId=id15136")
        await browser.close()
    return wc, prior, ranking, team_ids

wc_raw, prior_raw, ranking_raw, team_ids = asyncio.get_event_loop().run_until_complete(collect())
os.makedirs(DATA_DIR, exist_ok=True)

pd.DataFrame([{
    "MatchID": m["idMatchIfes"], "MatchDate": m["matchDate"],
    "TeamAID": m["teamAId"], "TeamBID": m["teamBId"],
    "TeamAGoals": m["teamAScore"], "TeamBGoals": m["teamBScore"],
    "TeamAPen": m["teamAPenaltyScore"], "TeamBPen": m["teamBPenaltyScore"],
    "HasPenalties": int(m["hasPenalties"]), "Stage": english(m["stageName"]),
    "Stadium": english(m["stadiumName"])} for m in wc_raw]
).to_csv(f"{DATA_DIR}/wc2026_matches.csv", index=False)

names = {}
for m in wc_raw:
    names[m["teamAId"]] = english(m["teamAName"]); names[m["teamBId"]] = english(m["teamBName"])
pd.DataFrame(list(names.items()), columns=["TeamID", "Team"]).to_csv(f"{DATA_DIR}/team_names.csv", index=False)

prior_df = pd.DataFrame([{
    "MatchID": m["idMatchIfes"], "MatchDate": m["matchDate"],
    "TeamAID": m["teamAId"], "TeamBID": m["teamBId"],
    "TeamAGoals": m["teamAScore"], "TeamBGoals": m["teamBScore"],
    "Competition": f'{english(m["competitionName"])}|{english(m["seasonName"])}'} for m in prior_raw])
prior_df = prior_df.drop_duplicates("MatchID")
prior_df = prior_df[(prior_df["MatchDate"] >= "2025-01-01") & (prior_df["MatchDate"] <= "2026-07-19")]
codes = {c: i for i, c in enumerate(prior_df["Competition"].unique())}
prior_df["CompCode"] = prior_df["Competition"].map(codes)
prior_df.drop(columns="Competition").to_csv(f"{DATA_DIR}/prior_matches.csv", index=False)
pd.DataFrame(list(codes.items()), columns=["Competition", "CompCode"])[["CompCode", "Competition"]] \
  .to_csv(f"{DATA_DIR}/competitions.csv", index=False)

pd.DataFrame([{
    "TeamID": r["rankingItem"]["idTeam"], "CountryCode": r["rankingItem"]["countryCode"],
    "Rank": r["rankingItem"]["rank"], "TotalPoints": r["rankingItem"]["totalPoints"],
    "PreviousRank": r["rankingItem"]["previousRank"], "PreviousPoints": r["previousPoints"]}
    for r in ranking_raw["rankings"] if r["rankingItem"]["idTeam"] in names]
).to_csv(f"{DATA_DIR}/fifa_ranking_2026_06_11.csv", index=False)
print("Raw data downloaded from FIFA.")
```

## Creating a Pool of Candidate Pre-Match Variables

Goals scored by a team depend on **its own attacking ability** and on **how well the opponent defends**, so candidate variables are created for both sides of each row:

| Group | Candidate variables |
|---|---|
| Team attack (recent form) | `TeamAvgGoalsScored5`, `TeamAvgGoalsScored10`, `TeamScoringRate5`, `TeamTwoPlusGoalsRate5`, `TeamWinRate5`, `TeamAvgGoalsConceded5` |
| Opponent defence (recent form) | `OppAvgGoalsConceded5`, `OppAvgGoalsConceded10`, `OppCleanSheetRate5`, `OppCleanSheetRate10`, `OppConcedeTwoPlusRate5`, `OppAvgGoalsScored5`, `OppWinRate5` |
| Overall strength | `TeamRankingPoints`, `OppRankingPoints`, `RankingPointsDifference` |
| Match context | `IsHostNation` (USA, Mexico, Canada), `OppIsHostNation`, `IsKnockout` (Round of 32 onwards), `TeamDaysRest`, `OppDaysRest` |

The stage of a match (group or knockout) and whether a team is a host nation are fixed by the tournament schedule, so they are known before kick-off.

```python
def recent_form(team_id, target_date, window):
    # Summarise a team's last `window` matches played strictly before target_date.
    previous = team_history_df[(team_history_df["TeamID"] == team_id) &
                               (team_history_df["MatchDate"] < target_date)].tail(window)
    if len(previous) < window:
        return None
    return {
        "AvgGoalsScored": previous["GoalsFor"].mean(),
        "AvgGoalsConceded": previous["GoalsAgainst"].mean(),
        "ScoringRate": (previous["GoalsFor"] > 0).mean(),
        "TwoPlusGoalsRate": (previous["GoalsFor"] >= 2).mean(),
        "WinRate": (previous["GoalsFor"] > previous["GoalsAgainst"]).mean(),
        "CleanSheetRate": (previous["GoalsAgainst"] == 0).mean(),
        "ConcedeTwoPlusRate": (previous["GoalsAgainst"] >= 2).mean(),
        "DaysRest": (target_date - previous["MatchDate"].max()).days,
    }

ranking_points = ranking.set_index("TeamID")["TotalPoints"]
HOST_NATIONS = {"USA", "Mexico", "Canada"}

candidate_rows = []
for _, row in team_match_df.iterrows():
    t5 = recent_form(row["TeamID"], row["MatchDate"], 5)
    t10 = recent_form(row["TeamID"], row["MatchDate"], 10)
    o5 = recent_form(row["OpponentID"], row["MatchDate"], 5)
    o10 = recent_form(row["OpponentID"], row["MatchDate"], 10)
    candidate_rows.append({
        "RowID": row["RowID"],
        "TeamAvgGoalsScored5": t5["AvgGoalsScored"],
        "TeamAvgGoalsScored10": t10["AvgGoalsScored"],
        "TeamScoringRate5": t5["ScoringRate"],
        "TeamTwoPlusGoalsRate5": t5["TwoPlusGoalsRate"],
        "TeamWinRate5": t5["WinRate"],
        "TeamAvgGoalsConceded5": t5["AvgGoalsConceded"],
        "OppAvgGoalsConceded5": o5["AvgGoalsConceded"],
        "OppAvgGoalsConceded10": o10["AvgGoalsConceded"],
        "OppCleanSheetRate5": o5["CleanSheetRate"],
        "OppCleanSheetRate10": o10["CleanSheetRate"],
        "OppConcedeTwoPlusRate5": o5["ConcedeTwoPlusRate"],
        "OppAvgGoalsScored5": o5["AvgGoalsScored"],
        "OppWinRate5": o5["WinRate"],
        "TeamRankingPoints": ranking_points[row["TeamID"]],
        "OppRankingPoints": ranking_points[row["OpponentID"]],
        "RankingPointsDifference": ranking_points[row["TeamID"]] - ranking_points[row["OpponentID"]],
        "IsHostNation": int(row["Team"] in HOST_NATIONS),
        "OppIsHostNation": int(row["Opponent"] in HOST_NATIONS),
        "IsKnockout": int(row["Stage"] != "First Stage"),
        "TeamDaysRest": t5["DaysRest"],
        "OppDaysRest": o5["DaysRest"],
    })

candidate_df = pd.DataFrame(candidate_rows)
candidate_columns = [c for c in candidate_df.columns if c != "RowID"]
candidate_analysis_df = team_match_df.merge(candidate_df, on="RowID", how="left")

print("Rows:", len(candidate_analysis_df))
print("Candidate variables:", len(candidate_columns))
print("Missing values:", candidate_analysis_df[candidate_columns].isna().sum().sum())
display(candidate_analysis_df.head())
```

## Selecting the Final Eight Explanatory Variables
The candidates were compared using their correlation with GoalsScored, their correlation with each other, and whether they make football sense for predicting **one team's** goals.
*Goals scored = team strength + team attacking form − opponent strength + opponent defensive weakness, adjusted for match context*.

**Exact or near-duplicate relationships removed**
- RankingPointsDifference = TeamRankingPoints − OppRankingPoints. All three cannot be used together (perfect multicollinearity). The two separate components are kept so that the model can weight the team's own strength and the opponent's strength differently, which suits a single-team response.
- Five- and ten-match versions of the same measure are highly correlated (TeamAvgGoalsScored5 vs TeamAvgGoalsScored10: r = 0.79; OppAvgGoalsConceded5 vs OppAvgGoalsConceded10: r = 0.83). The **ten-match** versions are kept because they correlate more strongly with goals and are more stable.
- OppCleanSheetRate10 is strongly correlated with OppAvgGoalsConceded10 (r = −0.80) and was removed.

**Variables that mostly repeat team strength removed**
- TeamWinRate5 (r = 0.7 with TeamRankingPoints), OppWinRate5 (r = 0.7 with OppRankingPoints) and OppAvgGoalsScored5 (r = 061 with OppRankingPoints) mainly measure overall strength, which ranking points already capture over a longer period.

**Weak variables removed**
- TeamScoringRate5 (r = 0.16), TeamAvgGoalsConceded5 (r = −0.15), OppCleanSheetRate5 (r = −0.14), OppIsHostNation (r = −0.07) and both rest-day variables (|r = -0.05 and 0.01).

**Variables retained**
- TeamTwoPlusGoalsRate5 and OppConcedeTwoPlusRate5 are kept alongside the ten-match averages. They describe short-term form and the ability to score (or concede) multiple goals.
- IsHostNation is weakly correlated (r = 0.10) but has a clear football rationale (home support, no long-haul travel). Host nations averaged 2.00 goals per match versus 1.44 for other teams, although only 15 rows are host-nation rows.
- IsKnockout is retained as a context variable to test the common belief that knockout matches are tighter.

### Final eight explanatory variables
1. TeamRankingPoints : team's FIFA ranking points (11 June 2026)
2. OppRankingPoints : opponent's FIFA ranking points (11 June 2026)

## Building the Linear Regression Model

**Keeping both rows of a match together**: Each match contributes two rows that share the same match conditions (and the two goal counts are related). If one row of a match were in the training set and the other in the test set, information about the test match would leak into training. Therefore the split is made **by match**. About 20% of matches are held out for testing.

**Tutorials:**
- [scikit-learn: GroupShuffleSplit](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GroupShuffleSplit.html)
- [scikit-learn: LinearRegression](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html)

```python
X = regression_2_2_df[selected_columns]
y = regression_2_2_df["GoalsScored"]
groups = regression_2_2_df["MatchID"]

splitter = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=RANDOM_STATE)
train_idx, test_idx = next(splitter.split(X, y, groups))
X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

print("Training rows:", len(X_train), "| matches:", groups.iloc[train_idx].nunique())
print("Testing rows:", len(X_test), "| matches:", groups.iloc[test_idx].nunique())
print("Matches in both sets:", len(set(groups.iloc[train_idx]) & set(groups.iloc[test_idx])))
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

## Evaluating the Model on the Test Set
The model is evaluated using MAE, RMSE and R².

**Tutorial:** [scikit-learn: Regression metrics](https://scikit-learn.org/stable/modules/model_evaluation.html#regression-metrics)

```python
def metrics(actual, predicted):
    return {"MAE": mean_absolute_error(actual, predicted),
            "RMSE": np.sqrt(mean_squared_error(actual, predicted)),
            "R2": r2_score(actual, predicted)}

test_results = pd.DataFrame({"Regression 2.2 model": metrics(y_test, y_pred)}).T
display(test_results.round(3))

lims = [-0.5, max(y_test.max(), y_pred.max()) + 0.5]
plt.figure(figsize=(7, 6))
plt.scatter(y_test + rng.uniform(-0.1, 0.1, len(y_test)), y_pred, alpha=0.7)
plt.plot(lims, lims, "--", color="black")
plt.xlabel("Actual goals scored")
plt.ylabel("Predicted goals scored")
plt.title("Actual vs Predicted Goals Scored (test set)")
plt.tight_layout()
plt.show()
```

- **Mean Absolute Error (MAE) of 0.99**
- **Root Mean Squared Error (RMSE) of 1.211**: This is slightly higher than the MAE, reflecting the typical influence of a few larger errors in matches with high goal counts.
- **R² of 0.234**: The model explains 23.4% of the variance in goals scored on this test set. While positive, the low R² suggests that match-day scoring remains highly unpredictable and is influenced heavily by factors beyond the historical team form and rankings captured in this model.

## Five-Fold Cross-Validation

A single 80/20 split depends on which matches happen to be in the test set. Five-fold cross-validation uses every match once for testing. GroupKFold keeps both rows of a match in the same fold. The same folds are used for the baseline so that the comparison is fair.
**Tutorials:**
- [scikit-learn: GroupKFold](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GroupKFold.html)
- [scikit-learn: cross_validate](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.cross_validate.html)


```python
from sklearn.dummy import DummyRegressor

gkf = GroupKFold(n_splits=5)
scoring = {"MAE": "neg_mean_absolute_error", "RMSE": "neg_root_mean_squared_error", "R2": "r2"}

def grouped_cv(estimator, X_, y_):
    res = cross_validate(estimator, X_, y_, groups=groups, cv=gkf, scoring=scoring)
    return pd.DataFrame({"MAE": -res["test_MAE"], "RMSE": -res["test_RMSE"], "R2": res["test_R2"]},
                        index=[f"Fold {i+1}" for i in range(5)])

cv_model = grouped_cv(LinearRegression(), X, y)
cv_baseline = grouped_cv(DummyRegressor(strategy="mean"), X, y)

display(cv_model.round(2))
cv_summary = pd.DataFrame({"Regression 2.2 model": cv_model.mean()}).T
cv_summary["R2 std"] = [cv_model["R2"].std()]
display(cv_summary.round(2))
```

Comparing the five-fold cross-validation metrics of the **Regression 2.2 model** against the initial **single train-test split** yields several key insights:
1. **A More Realistic and Stable Performance Estimate:**
   - **MAE:** The average cross-validation Mean Absolute Error of **0.96 goals** shows a slight improvement over the single test split's MAE of **0.99 goals**.
   - **RMSE:** Similarly, the average cross-validation RMSE (**1.26**) is close to the single split's RMSE of **1.21**, showing that overall error magnitudes remain fairly steady when evaluated across the entire dataset.
   - **$R^2$:** While the single train-test split yielded an $R^2$ of **0.234** (explaining 23.4% of the variance), the average cross-validation $R^2$ drops to **0.133** (13.3%).

2. **High Volatility and Sensitivity to Match Splitting:**
   - Evaluating the individual cross-validation folds reveals significant performance swings:
     - **Folds 1, 4, and 5** perform exceptionally well (producing positive $R^2$ values of **0.30**, **0.29**, and **0.26** respectively), which aligns with or exceeds the initial test split's $R^2$ of **0.234**.
     - **Folds 2 and 3** struggle considerably, yielding negative $R^2$ values of **-0.06** and **-0.13** respectively. This means that for these specific folds, a simple average prediction would have outperformed our regression model.
   - This variability is captured by the high standard deviation of the cross-validation $R^2$ scores (**0.21**). It highlights how sensitive our model's performance is to match-day anomalies and the specific subset of games in each fold (e.g., highly unexpected upsets, red cards, or unusual high-scoring matches which historical rankings and pre-match form cannot easily capture).

## Comparing Reasonable Alternatives

The selected model is compared with reasonable alternatives using the same grouped five-fold cross-validation:

1. **Alternative variable set (results-based form):** win rates replace the two-plus-goals rates.
2. **Alternative windows:** five-match averages replace the ten-match averages.
3. **Square-root transformation of the response:** goals are counts with a right skew; sqrt(GoalsScored) is a common variance-stabilising transformation. Predictions are back-transformed (squared) so that MAE and RMSE are on the goals scale.

```python
def cv_eval(columns, transform=None, drop_mask=None):
    Xa = candidate_analysis_df[columns].values; ya = y.values
    preds = np.zeros(len(ya))
    for tr, te in gkf.split(Xa, ya, groups):
        if drop_mask is not None:
            tr = tr[~drop_mask[tr]]
        m = LinearRegression()
        if transform == "sqrt":
            m.fit(Xa[tr], np.sqrt(ya[tr])); preds[te] = np.clip(m.predict(Xa[te]), 0, None) ** 2
        else:
            m.fit(Xa[tr], ya[tr]); preds[te] = m.predict(Xa[te])
    fold_r2 = [r2_score(ya[te], preds[te]) for _, te in gkf.split(Xa, ya, groups)]
    return {"MAE": mean_absolute_error(ya, preds), "RMSE": np.sqrt(mean_squared_error(ya, preds)),
            "Mean fold R2": np.mean(fold_r2)}

alt_results_based = ["TeamRankingPoints", "OppRankingPoints", "TeamAvgGoalsScored10", "OppAvgGoalsConceded10",
                     "TeamWinRate5", "OppWinRate5", "IsHostNation", "IsKnockout"]
alt_five_match = ["TeamRankingPoints", "OppRankingPoints", "TeamAvgGoalsScored5", "OppAvgGoalsConceded5",
                  "TeamTwoPlusGoalsRate5", "OppConcedeTwoPlusRate5", "IsHostNation", "IsKnockout"]

comparison = pd.DataFrame({
    "Selected model (8 variables)": cv_eval(selected_columns),
    "Alt 1: win rates instead of 2+ goal rates": cv_eval(alt_results_based),
    "Alt 2: 5-match averages": cv_eval(alt_five_match),
    "Alt 3: sqrt(GoalsScored) response": cv_eval(selected_columns, transform="sqrt"),
}).T
display(comparison.round(3))
```

Comparing the cross-validation performance of the selected model against the three candidate alternatives yields the following insights:

* **Alt 1 (Win rates instead of 2+ goal rates):** Replacing scoring consistency with results-based form yields a very slight improvement in Mean fold $R^2$ (from **0.133** to **0.145**) and marginally lowers the MAE. This suggests that a team's overall winning momentum might be a marginally cleaner signal of general quality than simply counting matches where they scored $2+$ goals.
* **Alt 2 (5-match averages):** Shortening the rolling historical window from 10 matches to 5 matches keeps the Mean fold $R^2$ practically identical (**0.133**). However, the longer 10-match baseline remains conceptually preferred as it smoothes out noisy, single-match statistical anomalies.
* **Alt 3 (Square-root transformation):** Applying a variance-stabilizing square-root transformation to the right-skewed GoalsScored response variable performs the worst. It results in a noticeable drop in Mean fold $R^2$ (**0.088**), indicating that back-transforming predictions (by squaring them) introduces non-linear projection distortions that degrade overall predictive accuracy on the goals scale.

## Conclusion: Predicting Goals Scored in FIFA World Cup 2026 Matches

Core analytic question: **Can pre-match information about a team and its opponent be used to predict the number of goals that a team scores in a FIFA World Cup 2026 match?**

### 1. Overall Predictive Ability
* **Moderate Explanatory Power:** The model yields an average cross-validation $R^2$ of **0.133** (~13.3% of variance explained), with single test-set performance reaching up to **23.4%** ($R^2 = 0.234$).
* **Baseline Accuracy:** The average Mean Absolute Error (MAE) of **0.96 goals** (cross-validation) and **0.99 goals** (test set) implies that, on average, predictions are off by just under one goal per team per match.

### 2. Key Predictors of Team Scoring
* **Opponent's Defensive Vulnerability:** Short-term defensive instability of the opponent (OppConcedeTwoPlusRate5, Coefficient: **~0.76**) is the single strongest positive predictor.
* **Home Advantage:** Host nation status (IsHostNation, Coefficient: **~0.50**) provides a massive boost of half a goal.
* **Long-Term vs. Short-Term Form:** An opponent's long-term leakiness (OppAvgGoalsConceded10, Coefficient: **~0.21**) exerts over twice the impact on predicted goals than a team's own long-term historical offensive average (TeamAvgGoalsScored10, Coefficient: **~0.10**).
* **Negligible Context Impacts:** Match context such as knockout-stage pressure (IsKnockout, Coefficient: **~-0.07**) shows a negligible effect on goal counts, debunking the common assumption that matches inherently become strictly defensive after the group stages.

### 3. Model Robustness & Limits of Predictability
* **High Volatility:** The high standard deviation in cross-validation $R^2$ scores (**0.21**) across folds highlights extreme sensitivity to match splits containing unexpected upsets, red cards, or injuries.
* **Right-Skewed Distribution:** Attempting a square-root transformation backfired, lowering the cross-validation $R^2$ to **0.088** due to non-linear distortion when back-transforming predictions.

### Final Answer
Yes, but only to a moderate extent. Pre-match information successfully isolates structural trends—chiefly opponent defensive instability, host-nation boosts, and long-term ranking gaps. However, approximately 80% of match-day goal variance remains driven by other unpredictable factors.
